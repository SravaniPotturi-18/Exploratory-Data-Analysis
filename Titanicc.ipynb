```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np
```

```python
sns.set_theme(style='whitegrid')
print("All libraries imported successfully")
```

```python
url = "https://raw.githubusercontent.com/datasciencedojo/datasets/master/titanic.csv"

df = pd.read_csv(url)
display(df.head())
```

```python
print("Number of Rows:", df.shape[0])
print("Number of Columns:", df.shape[1])

print("\nColumn Names:")
print(df.columns.tolist())

print("\nDataset Information:")
df.info()

print("\nStatistical Summary:")
display(df.describe())
```

```python
print("Missing Values in Each Column:")
print(df.isnull().sum())
```

```python
print("Number of Duplicate Rows:", df.duplicated().sum())
```

```python

print("Survival Counts:")
print(df['Survived'].value_counts())

print("\nPassenger Gender Counts:")
print(df['Sex'].value_counts())

print("\nPassenger Class Counts:")
print(df['Pclass'].value_counts())

print("\nEmbarkation Port Counts:")
print(df['Embarked'].value_counts())
```

```python
survival_rate = df['Survived'].mean() * 100

print("Overall Survival Rate:", round(survival_rate, 2), "%")
```

```python
plt.figure(figsize=(8, 5))

sns.histplot(data=df, x='Age', bins=20, kde=True)

plt.title('Distribution of Passenger Ages')
plt.xlabel('Age')
plt.ylabel('Number of Passengers')

plt.show()
```

```python
plt.figure(figsize=(8, 5))

sns.boxplot(data=df, x='Fare')

plt.title('Boxplot of Passenger Fares')
plt.xlabel('Fare')

plt.show()
```

```python
plt.figure(figsize=(8, 5))

sns.boxplot(data=df, x='Survived', y='Age')

plt.title('Age Distribution by Survival Status')
plt.xlabel('Survival Status (0 = No, 1 = Yes)')
plt.ylabel('Age')

plt.show()
```

```python
plt.figure(figsize=(8, 5))

sns.countplot(data=df, x='Sex', hue='Survived')

plt.title('Survival Count by Gender')
plt.xlabel('Gender')
plt.ylabel('Number of Passengers')
plt.legend(title='Survived', labels=['No', 'Yes'])

plt.show()
```

```python
plt.figure(figsize=(8, 5))

sns.countplot(data=df, x='Pclass', hue='Survived')

plt.title('Survival Count by Passenger Class')
plt.xlabel('Passenger Class')
plt.ylabel('Number of Passengers')
plt.legend(title='Survived', labels=['No', 'Yes'])

plt.show()
```

```python
plt.figure(figsize=(8, 5))

sns.scatterplot(
    data=df,
    x='Age',
    y='Fare',
    hue='Survived',
    alpha=0.7
)

plt.title('Passenger Age vs Fare')
plt.xlabel('Age')
plt.ylabel('Fare')
plt.legend(title='Survived')

plt.show()
```

```python
gender_survival = df.groupby('Sex')['Survived'].mean() * 100

print("Survival Rate by Gender:")
print(gender_survival.round(2))

gender_survival.plot(kind='bar', figsize=(7, 5))

plt.title('Survival Rate by Gender')
plt.xlabel('Gender')
plt.ylabel('Survival Rate (%)')
plt.xticks(rotation=0)

plt.show()
```

```python
numeric_df = df.select_dtypes(include='number')

correlation_matrix = numeric_df.corr()

plt.figure(figsize=(10, 7))

sns.heatmap(
    correlation_matrix,
    annot=True,
    cmap='coolwarm',
    fmt='.2f',
    linewidths=0.5
)

plt.title('Correlation Heatmap of Titanic Dataset')

plt.show()
```

```python
pairplot_data = df[
    ['Survived', 'Pclass', 'Age', 'Fare', 'SibSp', 'Parch']
].dropna()

sns.pairplot(
    pairplot_data,
    hue='Survived',
    diag_kind='hist',
    corner=True
)

plt.suptitle(
    'Pairplot of Titanic Numerical Variables',
    y=1.02
)

plt.show()
```

```python
from statsmodels.stats.outliers_influence import variance_inflation_factor

vif_data = df[
    ['Pclass', 'Age', 'SibSp', 'Parch', 'Fare']
].copy()

# Fill missing numerical values with their column medians
vif_data = vif_data.fillna(vif_data.median())

vif_results = pd.DataFrame()
vif_results['Feature'] = vif_data.columns
vif_results['VIF'] = [
    variance_inflation_factor(vif_data.values, i)
    for i in range(vif_data.shape[1])
]

display(vif_results.round(2))
```

```python
print("EXPLORATORY DATA ANALYSIS - KEY FINDINGS")
print("=" * 45)

# Overall survival rate
survival_rate = df['Survived'].mean() * 100
print(f"\n1. Overall Survival Rate: {survival_rate:.2f}%")

# Survival rate by gender
gender_rate = df.groupby('Sex')['Survived'].mean() * 100
print("\n2. Survival Rate by Gender:")
print(gender_rate.round(2))

# Survival rate by passenger class
class_rate = df.groupby('Pclass')['Survived'].mean() * 100
print("\n3. Survival Rate by Passenger Class:")
print(class_rate.round(2))

# Average fare by passenger class
fare_by_class = df.groupby('Pclass')['Fare'].mean()
print("\n4. Average Fare by Passenger Class:")
print(fare_by_class.round(2))

# Missing values
print("\n5. Columns with Missing Values:")
print(df.isnull().sum()[df.isnull().sum() > 0])

# Age statistics
print("\n6. Passenger Age Statistics:")
print(df['Age'].describe().round(2))
```

## Conclusion

Exploratory Data Analysis was performed on the Titanic passenger dataset using Python libraries, including Pandas, NumPy, Matplotlib, and Seaborn.

The analysis included dataset inspection, statistical summaries, missing-value identification, duplicate checks, histograms, boxplots, bar charts, scatterplots, a correlation heatmap, and a pairplot.

The findings indicate that survival outcomes varied across passenger gender and class. The visualizations also revealed differences in age and fare distributions, potential fare outliers, and missing values in selected columns.

Correlation analysis helped explore relationships between numerical variables, while the Variance Inflation Factor (VIF) was used to assess potential multicollinearity among selected predictors.

Overall, this project demonstrates how Exploratory Data Analysis can help identify patterns, trends, data-quality issues, and relationships before further statistical modelling. The findings describe associations in the dataset and do not establish causation.
